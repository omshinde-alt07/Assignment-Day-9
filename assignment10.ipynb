Q1 — Tuple Immutability Trap

```python
t = ([1,2], [3,4])
t[0][0] = 99
```

Does it work?
Yes, it works.

Result
t = ([99,2], [3,4])
Explanation

A tuple is immutable, meaning:
You cannot change the tuple itself
You cannot reassign elements of the tuple

So when we do:
t[0][0] = 99
We are modifying the list inside the tuple, not the tuple structure.

Q2 — Duplicate Detection (O(n) using sets)

```python
def find_duplicates(lst):
    seen = set()
    duplicates = set()

    for item in lst:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)

    return duplicates

find_duplicates([1,2,3,2,4,5,3,6])
```

Q3 — Debug Problem

```python
def unique_to_each(a, b):
    result = set(a) - set(b)
    return list(result)

unique_to_each([1,2,3], [3,4,5])
```

Why This Happens

The code only computes:

set(a) - set(b)

Which means:

elements in a but not in b

{1,2,3} - {3,4,5}
= {1,2}

But the question asks for:

elements unique to each list

So we need:

(a - b)  +  (b - a)

This is exactly the symmetric difference operation.

Correct Solution
Method 1 — Using symmetric_difference

```python
def unique_to_each(a, b):
    return list(set(a).symmetric_difference(set(b)))
```

Method 2 — Using operator ^

```python
def unique_to_each(a, b):
    return list(set(a) ^ set(b))
```
